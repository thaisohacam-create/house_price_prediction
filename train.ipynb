```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns
```

```python
data=pd.read_csv(r'C:\Users\Qin\Desktop\CODE\sale_real_estate.csv')
data['address'] = data['address'].str.replace('·\n','',regex=False)
data1=data.copy()
data1
```

```python
print(data.duplicated().sum())
data1=data.drop_duplicates()
data1.info()
```

```python
address_stats=data.groupby('address')['address'].agg('count').sort_values(ascending=False)
address_stats
```

```python
len(address_stats[address_stats<10])
```

```python
address_stats_less_than_10=address_stats[address_stats<10]
address_stats_less_than_10
```

```python
data1.address=data1.address.apply(lambda x:'other' if x in address_stats_less_than_10 else x)
```

```python

len(data1['price'].astype(str).str.isalpha())
data1
```

```python
data1['price']=data1['price'].str.replace('.','').str.replace(',','.')
```

```python
def convert_number(price):
    if pd.isna(price):
        return None
    price=price.strip()
    if 'tỉ' in price or 'tỷ' in price:
        price=price.replace('tỉ','').replace('tỷ','')
        return float(price.strip().replace(',', '.')) * 1_000_000_000
    elif 'triệu' in price:
        price=price.replace('triệu','')
        return float(price.replace('triệu',''))*1000000
data1['price']=data1['price'].apply(convert_number)
data1
```

```python
data1['area']=data1['area'].str.replace('m²','').str.replace('.','').str.replace(',','.')
data1['area']=data1['area'].astype(float)
data1
```

```python
data1['price_per_sqmt']=data1['price']/data1['area']
data1
```

```python
data1.price_per_sqmt.describe()
```

```python
def remove_pps_outliers(data1):
    data_out=pd.DataFrame()
    for key, subdata in data1.groupby('area'):
        m=np.mean(subdata.price_per_sqmt)
        st=np.std(subdata.price_per_sqmt)
        reduced_data=subdata[(subdata.price_per_sqmt>(m-st))&(subdata.price_per_sqmt<=(m+st))]
        data_out=pd.concat([data_out,reduced_data],ignore_index=True)
    return data_out
data1=remove_pps_outliers(data1)
data1.shape
```
